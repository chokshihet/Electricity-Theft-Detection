## Section 7 — Investigating the 24-hour block hypothesis

The reference paper describes theft generation using daily consumption profiles consisting of 24 hourly readings.

Based on this, we hypothesize that every 24 consecutive records in our dataset represent one daily consumption profile.

We will investigate the distribution of theft labels within these 24-record blocks to understand how individual hourly labels can be converted into a single daily target label.


### 7.1 Dataset Preparation

Before grouping the records into 24-hour blocks, remove the 15 invalid records with `Class = 0` identified during the initial data-quality analysis.


```python

rows_before = len(df)

df = df[df["Class"].astype(str).str.strip() != "0"].reset_index(drop=True)

print("Rows removed:", rows_before - len(df))
print("Remaining rows:", len(df))
print("Can be divided into complete 24-hour blocks:", len(df) % 24 == 0)
```


### 7.2 Distribution of Records Across Building Types

Examine the number of hourly observations available for each building type and check whether these counts are divisible by 24.


```python

records_per_class = df["Class"].value_counts().sort_index()

class_summary = pd.DataFrame({
    "Total Records": records_per_class,
    "24-Record Blocks": records_per_class // 24,
    "Remainder": records_per_class % 24
})

display(class_summary)
```


### 7.3 Building-Type Consistency Within Consecutive Blocks

Group the dataset into consecutive blocks of 24 records and check whether each block contains observations from only one building type.

This helps determine whether the assumed block size maintains building-type consistency.


```python

block_id = np.arange(len(df)) // 24

class_counts_per_block = df.groupby(block_id)["Class"].nunique()

print("Total blocks:", len(class_counts_per_block))
print("Blocks containing one building type:",
      (class_counts_per_block == 1).sum())
print("Blocks containing multiple building types:",
      (class_counts_per_block > 1).sum())
```


### 7.4 Verify Block Sizes

Finally, verify that every consecutive block contains exactly 24 records, ensuring that no incomplete blocks are introduced during the grouping process.


```python

block_sizes = df.groupby(block_id).size()

print("Minimum block size:", block_sizes.min())
print("Maximum block size:", block_sizes.max())
print("Blocks containing exactly 24 records:",
      (block_sizes == 24).sum())
print("Total blocks:", len(block_sizes))
```


### 7.5 Analysis of Consecutive Theft-Label Runs

Before grouping records into 24-hour blocks, investigate the consecutive runs of identical theft labels.

A run is a sequence of consecutive records having the same theft label. If the dataset is organized into 24-hour profiles, we expect the boundaries between daily profiles to influence the lengths of these runs.

We identify runs whose lengths are not divisible by 24 and examine whether their combined length is divisible by 24.


```python

# Identify consecutive runs of identical theft labels
labels = df["theft"].to_numpy()

change_indices = np.where(labels[1:] != labels[:-1])[0] + 1

starts = np.r_[0, change_indices]
ends = np.r_[change_indices, len(labels)]

# Create a DataFrame containing run information
runs = pd.DataFrame({
    "start_row": starts,
    "end_row": ends - 1,
    "theft": labels[starts],
    "run_length": ends - starts
})

# Separate runs divisible and not divisible by 24
divisible_runs = runs[runs["run_length"] % 24 == 0]
non_divisible_runs = runs[runs["run_length"] % 24 != 0]

print("Total consecutive runs:", len(runs))
print("Runs divisible by 24:", len(divisible_runs))
print("Runs NOT divisible by 24:", len(non_divisible_runs))

print("\nFirst 20 runs not divisible by 24:")
display(non_divisible_runs.head(20))
```


### 7.6 Combined Length of Non-Divisible Runs

Calculate the total number of records belonging to runs whose individual lengths are not divisible by 24.

Check whether this combined count is divisible by 24. This helps investigate whether these shorter or irregular runs collectively account for complete 24-record groups.


```python

total_non_divisible_rows = non_divisible_runs["run_length"].sum()

print("Number of non-divisible runs:", len(non_divisible_runs))
print("Combined number of records:", total_non_divisible_rows)
print("Remainder after division by 24:", total_non_divisible_rows % 24)
print("Is the combined length divisible by 24:",
      total_non_divisible_rows % 24 == 0)
```


### 7.6 Number of Distinct Theft Categories Per Block

Count the distinct theft categories occurring within each 24-record block.

`Normal` is excluded from this count because a block containing Normal records and one theft category should still be considered a single-theft-category block.


```python

theft_only = df["theft"].where(df["theft"] != "Normal")

theft_categories_per_block = (
    theft_only.groupby(block_id).nunique()
)

block_summary = (
    theft_categories_per_block.value_counts()
    .sort_index()
)

print("Distribution of distinct theft categories per block:")
display(block_summary)

print("\nTotal blocks:", len(theft_categories_per_block))
```


### 7.7 Investigation of Blocks Containing Multiple Theft Categories

Inspect blocks containing more than one distinct theft category to identify which categories occur together and determine whether a consistent pattern exists.


```python

mixed_block_ids = theft_categories_per_block[
    theft_categories_per_block > 1
].index

mixed_blocks = (
    df.assign(block_id=block_id)
      .loc[lambda x: x["block_id"].isin(mixed_block_ids)]
      .groupby(["block_id", "theft"])
      .size()
      .unstack(fill_value=0)
)

display(mixed_blocks)

mixed_labels = (
    df.assign(block_id=block_id)
      .loc[lambda x: x["block_id"].isin(mixed_block_ids)]
      .groupby("block_id")["theft"]
      .apply(lambda x: set(x) - {"Normal"})
)

all_mixed_include_theft2 = mixed_labels.apply(
    lambda x: len(x) == 2 and "Theft2" in x
).all()

print("Number of mixed blocks:", len(mixed_labels))
print("Every mixed block contains Theft2 and exactly one other theft category:",
      all_mixed_include_theft2)
```


### 7.8 Conclusion and Block-Level Label Assignment

The investigation revealed three possible patterns within a 24-record block:

1. All 24 records are labelled Normal.
2. Records contain Normal and exactly one theft category.
3. Records contain two distinct theft categories.

Further inspection showed that every block containing two theft categories included Theft2 along with exactly one other theft category.

Based on these observations, the following labelling rules are adopted:

- All records are Normal → Normal
- Normal and one theft category → Assign that theft category
- Two theft categories, one of which is Theft2 → Assign Theft2

These rules allow each 24-hour block to be represented by a single target label.


## Transforming and Encoding Data

```python
# Identify consumption feature columns
feature_cols = [
    col for col in df.columns
    if col not in ["0", "Class", "theft"]
]

# Check that all rows can be divided into complete 24-hour blocks
if len(df) % 24 != 0:
    raise ValueError("Dataset rows are not divisible by 24.")

# Check for consistent building type within each block
class_counts = df.groupby(np.arange(len(df)) // 24)["Class"].nunique()

if (class_counts > 1).any():
    raise ValueError("A 24-hour block contains multiple building types.")

# Transform data into daily samples
transformed_rows = []

for start in range(0, len(df), 24):

    block = df.iloc[start:start + 24]

    row = {}

    # Preserve building type
    row["Class"] = block["Class"].iloc[0]

    # Convert 24 hourly readings into separate features
    for feature in feature_cols:
        for hour in range(24):
            row[f"{feature}_h{hour:02d}"] = block[feature].iloc[hour]

    # Determine target label
    theft_categories = set(block["theft"].unique()) - {"Normal"}

    if len(theft_categories) == 0:
        label = "Normal"

    elif len(theft_categories) == 1:
        label = next(iter(theft_categories))

    elif len(theft_categories) == 2:
        if "Theft2" in theft_categories:
            label = "Theft2"
        else:
            raise ValueError(
                f"Unexpected theft combination: {theft_categories}"
            )

    else:
        raise ValueError(
            f"Unexpected number of theft categories: {theft_categories}"
        )

    row["theft"] = label

    transformed_rows.append(row)

# Create transformed DataFrame
transformed_df = pd.DataFrame(transformed_rows)

# Save to CSV
transformed_df.to_csv("../data/transformed_data.csv", index=False)

# Display results
print("Transformation completed successfully!")
print("Shape:", transformed_df.shape)

print("\nTarget distribution:")
print(transformed_df["theft"].value_counts())

print("\nFirst few feature names:")
print(transformed_df.columns[:12].tolist())

print("\nSaved as transformed_data.csv")
```

```python
## One hot encoding for class(Building type)
# Load transformed dataset
transformed_df = pd.read_csv("../data/transformed_data.csv")

# One-hot encode building type
encoded_df = pd.get_dummies(transformed_df, columns=["Class"], dtype=int)

# Separate target for clarity
X = encoded_df.drop(columns=["theft"])
y = encoded_df["theft"]

# Save encoded dataset
encoded_df.to_csv("../data/encoded_data.csv", index=False)

print("Encoding completed!")
print("Dataset shape:", encoded_df.shape)

print("\nTarget classes:")
print(y.value_counts())

print("\nEncoded Class columns:")
print([col for col in encoded_df.columns if col.startswith("Class_")])
```
