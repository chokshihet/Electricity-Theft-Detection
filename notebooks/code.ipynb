### 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
pd.set_option('display.max_columns', None)
```

### 2. Load the dataset

```python
df = pd.read_csv("../data/df.csv")

print("Dataset Shape:", df.shape)
display(df.head())
```

### 3. Understanding the dataset

```python
print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nStatistical Summary:")
display(df.describe(include="all").T)
```

### 4. Analyzing theft categories

```python
print("Theft Category Distribution:")
display(df["theft"].value_counts())

plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x="theft",
    order=["Normal", "Theft1", "Theft2", "Theft3", "Theft4", "Theft5", "Theft6"]
)

plt.title("Distribution of Theft Categories")
plt.xlabel("Theft Category")
plt.ylabel("Number of Records")
plt.show()
```

### 5. Analyzing the buildging categories

```python
print("Building Category Distribution:")
display(df["Class"].value_counts())

plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    y="Class",
    order=df["Class"].value_counts().index
)

plt.title("Distribution of Building Categories")
plt.xlabel("Number of Records")
plt.ylabel("Building Category")
plt.show()
```

### 6. Data Quality Analysis

#### 6.1 Investigation of Invalid Building Categories

```python
invalid_records = df[df["Class"].astype(str) == "0"]
print("Number of Invalid Building Records: ", len(invalid_records))

display(invalid_records)

print("\nTheft Distribution of Invalid Records:")
print(invalid_records["theft"].value_counts())
```

#### 6.2 Check duplicate records correctly

```python
print("Duplicates including Row Identifiers:")
print(df.duplicated().sum())

print("\nDuplicates Excluding Row Identifiers: ")
print(df.drop(columns=["0"]).duplicated().sum())

print("\nDuplicate Percentage: ")
duplicate_count = df.drop(columns=["0"]).duplicated().sum()
duplicate_percentage = (duplicate_count / len(df)) * 100

print(f"{duplicate_percentage:.2f}%")
```

#### 6.3 Check negative consumption values

```python
numeric_columns = df.select_dtypes(include=["number"]).columns.drop("0")

negative_counts = (df[numeric_columns] < 0).sum()

print("Negative Values in Each Column:")
display(negative_counts)

print("\nTotal Negative Values: ", negative_counts.sum())
```

#### 6.4 Check Zero Consumption Values

```python
zero_counts = (df[numeric_columns] == 0).sum()

zero_percentage = (zero_counts / len(df)) * 100

zero_summary = pd.DataFrame({
    "Zero Count" : zero_counts,
    "Zero Percentage (%)" : zero_percentage.round(2)
})

display(zero_summary)
```

#### 6.5 Investigate the Theft2 pattern

```python
columns_to_check = [
    "Electricity:Facility [kW](Hourly)",
    "InteriorLights:Electricity [kW](Hourly)",
    "InteriorEquipment:Electricity [kW](Hourly)",
]

for col in columns_to_check:
    print(f"\n{col}")

    zero_by_theft = (
        df.groupby("theft")[col]
        .apply(lambda x: (x == 0).sum())
    )

    print("Zero Values by Theft Category:")
    display(zero_by_theft)
```
